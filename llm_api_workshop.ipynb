<p>مدل زبانی بزرگ (Large Language Model) برنامه‌ای است که می‌تواند ورودی متنی را پردازش کند و پاسخ متنی یا خروجی‌های دیگر تولید کند. وقتی از وب‌سایت یک چت‌بات استفاده می‌کنیم، رابط کاربری و بخش زیادی از ارتباط با مدل را سرویس‌دهنده مدیریت می‌کند.</p>
<p>اما وقتی از <strong>API</strong> استفاده می‌کنیم، برنامهٔ خودمان مستقیماً درخواست را به سرویس مدل می‌فرستد و پاسخ را دریافت می‌کند. در نتیجه می‌توانیم مدل را داخل یک وب‌سایت، ربات، ابزار آموزشی یا برنامهٔ پایتونی قرار دهیم.</p>
<h3>مسیر سادهٔ یک درخواست</h3>
<ol>
<li>کاربر در برنامهٔ ما یک پرسش وارد می‌کند.</li>
<li>کد Python درخواست را با کمک کتابخانهٔ <code>openai</code> آماده می‌کند.</li>
<li>درخواست به API سرویس دهنده فرستاده می شود.</li>
<li>مدل درخواست را پردازش می‌کند.</li>
<li>پاسخ به برنامهٔ ما برمی‌گردد و ما آن را نمایش می‌دهیم یا در منطق برنامه استفاده می‌کنیم.</li>
</ol>

<h3>خواندن تنظیمات از فایل <code>.env</code></h3>
<p>تابع <code>load_dotenv</code> مقادیر فایل را به متغیرهای محیطی اضافه می‌کند. سپس با <code>os.getenv</code> مقدار هر متغیر را می‌خوانیم.</p>

````python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("API_KEY")
BASE_URL = os.getenv("BASE_URL")
MODEL = os.getenv("MODEL")

if not API_KEY:
    raise RuntimeError(
        "API_KEY is missing. Please set it in your .env file."
    )

if not BASE_URL:
    raise RuntimeError(
        "BASE_URL is missing. Please set it in your .env file "
        "(e.g. https://api.openai.com/v1)."
    )

if not MODEL:
    raise RuntimeError(
        "MODEL is missing. Please set it in your .env file "
        "(e.g. gpt-4o-mini)."
    )

print("Configuration loaded successfully. You're ready to go.")
````

<h3>ساخت Client</h3>
<p>شیء <code>client</code> واسطهٔ ارتباط ما با API است. پس از ساخت آن، متدهای SDK را برای ارسال درخواست استفاده می‌کنیم.</p>
<p>در صورت استفاده از یک درگاه سازگار، <code>base_url</code> به آدرس API آن درگاه اشاره می‌کند. سازگاری فقط به داشتن نام «OpenAI-compatible» محدود نیست؛ سرویس‌دهنده باید endpoint و قابلیت مورد استفاده، از جمله Chat Completions، را واقعاً پشتیبانی کند.</p>

````python
from openai import OpenAI

client = OpenAI(
    base_url="https://api.gapgpt.app/v1",
    api_key=API_KEY
)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h2>۴. اولین درخواست به مدل</h2>
<p>درخواست سادهٔ ما سه قسمت مهم دارد:</p>
<ul>
<li><strong>model:</strong> مشخص می‌کند از کدام مدل استفاده شود.</li>
<li><strong>messages:</strong> متن یا ورودی‌ای که می‌خواهیم مدل پردازش کند.</li>
<li><strong>max_completion_tokens:</strong> حداکثر تعداد توکن خروجی</li>
</ul>
<p>در این کارگاه از <strong>Chat Completions API</strong> استفاده می‌کنیم و ورودی را فقط با لیست <code>messages</code> برای مدل می‌فرستیم.</p>
</div>

````python
response = client.chat.completions.create(
    model=MODEL,
    messages = [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is Python?"},
    ],
    # max_completion_tokens=250,
)

result = response.choices[0].message.content
print(result)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>کد را باز کنیم</h3>

<li><code>max_completion_tokens</code> سقف توکن‌های خروجی را مشخص می‌کند؛ این مقدار به معنی تضمین تعداد دقیق کلمات نیست.</li>

````python
if response.usage is not None:
    print("Input token: ", response.usage.prompt_tokens)
    print("Output token: ", response.usage.completion_tokens)
    print("Total: ", response.usage.total_tokens)
else:
    print("There not any info!")
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h2>۵. System Prompt و User Prompt</h2>
<p>مدل فقط متن سؤال را دریافت نمی‌کند. ما می‌توانیم به آن دستورالعمل‌هایی بدهیم که نقش، سبک، هدف و محدودیت‌های پاسخ‌گویی را تعیین کنند.</p>
<h3>System Prompt چیست؟</h3>
<p>به‌صورت مفهومی، System Prompt دستورالعمل نسبتاً پایداری است که می‌گوید دستیار چه نقشی دارد و پاسخ‌ها را با چه قواعدی تولید کند. مثلاً:</p>
<ul>
<li>نقش: «تو یک مربی برنامه‌نویسی هستی.»</li>
<li>مخاطب: «برای فردی توضیح بده که تازه Python را شروع کرده است.»</li>
<li>سبک: «با مثال کوتاه و فارسی روان توضیح بده.»</li>
<li>قالب: «پاسخ را در سه بخش تعریف، مثال و جمع‌بندی بنویس.»</li>
<li>محدودیت: «اگر از پاسخ مطمئن نیستی، حدس نزن و عدم اطمینان را بیان کن.»</li>
</ul>
<h3>User Prompt چیست؟</h3>
<p>درخواست یا داده‌ای است که کاربر در همان نوبت ارائه می‌کند؛ مثلاً «حلقهٔ for را با یک مثال توضیح بده.»</p>
<p><strong>تشبیه:</strong> System Prompt شبیه شرح وظایف و قواعد کاری یک همکار است؛ User Prompt کاری است که همین الان از او می‌خواهیم انجام دهد.</p>
</div>

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>System Prompt در Chat Completions کجا قرار می‌گیرد؟</h3>
<p>در مثال‌های این جزوه، دستورالعمل سطح بالا را در پیامی با نقش <code>system</code> می‌گذاریم و درخواست جاری کاربر را در پیامی با نقش <code>user</code>؛ هر دو داخل لیست <code>messages</code> قرار می‌گیرند.</p>
<p>در APIهای جدیدتر ممکن است با نقش <code>developer</code> هم روبه‌رو شوی. مهم این است که سطح دستورالعمل برنامه را با درخواست عادی کاربر قاطی نکنی.</p>
</div>

````python
SYSTEM_PROMPT = """
You are a Python teacher.
Your student is new to programming.
Use simple Persian and small examples.
If a question is not clear, ask first.
If you don't know the answer, don't guess.
""".strip()

USER_PROMPT = "What is the for loops in python?"

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT},
    ],
    max_completion_tokens=300,
)

result = response.choices[0].message.content

print(result)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>یک سؤال ثابت، سه دستورالعمل متفاوت</h3>
<p>برای اینکه اثر System Prompt ملموس شود، سؤال را ثابت نگه می‌داریم و فقط دستورالعمل را تغییر می‌دهیم. تفاوت خروجی‌ها را با هم مقایسه کنید.</p>
</div>

````python

question = "What is python?"

prompt_variants = {
    "For beginners": "Explain it to someone who is new to the topic. Use a simple example.",
    
    "For technical students": "Explain it clearly. Introduce the main terms and give a short technical example.",
    
    "In one sentence": "Explain it in one short sentence.",
}

for title, instruction in prompt_variants.items():
    result = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": instruction},
            {"role": "user", "content": question},
        ],
        max_completion_tokens=500,
    )

    print(f"\n--- {title} ---")

    message = result.choices[0].message

    print("Finish reason:", result.choices[0].finish_reason)
    print("Content:", message.content)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<p><strong>چه چیزی را مقایسه کنیم؟</strong> سطح جزئیات، لحن، طول پاسخ، واژگان تخصصی و نوع مثال. دقت کن که دستورالعمل جهت‌دهنده است، نه یک تضمین ریاضی برای خروجی کاملاً یکسان در همهٔ اجراها.</p>


<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>الگوی ساخت یک System Prompt مناسب</h3>
<p>برای نوشتن دستورالعمل بهتر، این سؤال‌ها را پاسخ بده:</p>
<ol>
<li><strong>نقش:</strong> مدل قرار است چه نقشی داشته باشد؟</li>
<li><strong>مخاطب:</strong> برای چه کسی پاسخ می‌دهد؟</li>
<li><strong>وظیفه:</strong> دقیقاً چه کاری باید انجام دهد؟</li>
<li><strong>قالب:</strong> پاسخ در چه شکل و چه اندازه‌ای باشد؟</li>
<li><strong>محدودیت:</strong> در چه شرایطی باید سؤال بپرسد یا عدم اطمینان را اعلام کند؟</li>
</ol>
<p>در صورت نیاز می‌توان نمونهٔ ورودی و خروجی مطلوب هم اضافه کرد. به این روش معمولاً Few-shot Prompting گفته می‌شود.</p>
</div>

````python
SYSTEM_PROMPT_TEMPLATE = """
نقش: تو یک دستیار آموزشی برای درس {subject} هستی.
مخاطب: دانش‌آموز مبتدی.
وظیفه: مفاهیم را دقیق و مرحله‌به‌مرحله توضیح بده.
سبک: فارسی روان، جمله‌های کوتاه و مثال‌های ساده.
قالب پاسخ:
1) تعریف کوتاه
2) مثال
3) جمع‌بندی
محدودیت: اگر اطلاعات کافی نداری، صریح بگو.
""".strip()

my_system_prompt = SYSTEM_PROMPT_TEMPLATE.format(subject="هوش مصنوعی")
print(my_system_prompt)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h2>۶. آیا مدل مکالمه‌های قبلی را به خاطر می‌سپارد؟</h2>
<p>یک درخواست مستقل به API معمولاً به‌تنهایی از درخواست قبلی ما آگاه نیست. اگر می‌خواهیم مدل از صحبت‌های قبلی استفاده کند، باید تاریخچه را دوباره در ورودی بفرستیم یا از امکانات مدیریت وضعیت مکالمهٔ سرویس استفاده کنیم.</p>
<p>برای یادگیری، در این جلسه تاریخچه را خودمان با یک <code>list</code> ساده مدیریت می‌کنیم. این کار نشان می‌دهد «حافظهٔ چت‌بات» در ساده‌ترین حالت چگونه ساخته می‌شود.</p>
</div>

````python

conversation_history = []

# First message
user_message = "My name is Sara."

conversation_history.append(
    {"role": "user", "content": user_message}
)

# Send message to AI
response = client.chat.completions.create(
    model=MODEL,
    messages=conversation_history
)

ai_message = response.choices[0].message.content

# Save AI response
conversation_history.append(
    {"role": "assistant", "content": ai_message}
)

print("AI:", ai_message)


# Second message
user_message = "I am learning Python."

conversation_history.append(
    {"role": "user", "content": user_message}
)

response = client.chat.completions.create(
    model=MODEL,
    messages=conversation_history
)

ai_message = response.choices[0].message.content

conversation_history.append(
    {"role": "assistant", "content": ai_message}
)

print("AI:", ai_message)


# Third message
user_message = "What is my name and what am I learning?"

conversation_history.append(
    {"role": "user", "content": user_message}
)

response = client.chat.completions.create(
    model=MODEL,
    messages=conversation_history
)

ai_message = response.choices[0].message.content

conversation_history.append(
    {"role": "assistant", "content": ai_message}
)

print("AI:", ai_message)


# Show chat history
print("\nChat History:")

for message in conversation_history:
    print(message["role"], ":", message["content"])
````

<h3>تبدیل این ایده به یک تابع</h3>
<p>هر بار پیام جدید کاربر را به تاریخچه اضافه می‌کنیم، تاریخچه را برای مدل می‌فرستیم و سپس پاسخ مدل را نیز ذخیره می‌کنیم.</p>
</div>

````python

SYSTEM_PROMPT = """
You are a friendly Python teacher.
Your student is a beginner.
Use simple Persian.
Explain step by step.
Give small Python examples.
If you don't know, say so.
""".strip()

chat_history = []

def ask_assistant(user_text, history):
    """Send a message and save the conversation."""

    history.append({"role": "user", "content": user_text})

    result = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            *history, # Add chat history to messages
        ],
        max_completion_tokens=350,
    )

    answer = result.choices[0].message.content.strip()
    history.append({"role": "assistant", "content": answer})

    return answer


# First question
print("User: What is a variable?")
print("AI:", ask_assistant("What is a variable?", chat_history))

# Second question
print("\nUser: How can I create one?")
print("AI:", ask_assistant("How can I create one?", chat_history))

# Third question
print("\nUser: Give me a simple example.")
print("AI:", ask_assistant("Give me a simple example.", chat_history))

# Fourth question
print("\nUser: What was my first question?")
print("AI:", ask_assistant("What was my first question?", chat_history))


# Show chat history
print("\nSaved messages:", len(chat_history))
````

<h2>۷. Streaming؛ دریافت پاسخ به‌صورت تدریجی</h2>
<p>در حالت عادی، برنامه منتظر می‌ماند تا پاسخ کامل تولید شود و سپس آن را دریافت می‌کند. در <strong>Streaming</strong> بخش‌های پاسخ به‌تدریج به برنامه می‌رسند و ما می‌توانیم همان لحظه آن‌ها را نمایش دهیم.</p>
<p>این قابلیت برای چت‌بات‌ها تجربهٔ کاربری بهتری ایجاد می‌کند؛ کاربر لازم نیست تا پایان تولید پاسخ منتظر بماند تا اولین متن را ببیند.</p>
<p>در Chat Completions پاسخ به‌صورت تکه‌هایی (chunk) می‌رسد. برای نمایش تدریجی متن، فیلد <code>delta.content</code> را در هر chunk بررسی می‌کنیم.</p>
</div>

````python

import time
stream = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a Python teacher. Use simple English."},
        {"role": "user", "content": "What is a function in Python? Give an example."},
    ],
    stream=True,
)

full_answer = ""

for chunk in stream:
    text = chunk.choices[0].delta.content

    if text:
        print(text, end="", flush=True)
        full_answer += text
    time.sleep(0.05)

print("\n\nFull answer saved!")
````


<div dir="rtl" style="text-align:right; line-height:1.9;">
    <h3>کد Streaming را بفهمیم</h3>
    <ul>
        <li>با <code>stream=True</code> از API می‌خواهیم پاسخ را به‌صورت تدریجی ارسال کند.</li>
        <li>با حلقه <code>for</code>، بخش‌های پاسخ (chunk) را یکی‌یکی دریافت می‌کنیم.</li>
        <li>با <code>delta.content</code> متن هر بخش را استخراج می‌کنیم.</li>
        <li>با <code>if text</code> بررسی می‌کنیم که بخش دریافتی خالی نباشد.</li>
        <li>با <code>print(text, end="", flush=True)</code> هر بخش را بلافاصله نمایش می‌دهیم.</li>
    </ul>
</div>


<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>نمایش پاسخ با Markdown در Jupyter</h3>
<p>کتابخانه‌ای که برای نمایش Markdown در Notebook به کارمان می‌آید، <code>IPython.display</code> است. با <code>Markdown</code> متن را به قالب Markdown تبدیل می‌کنیم و با <code>display</code> آن را در خروجی Jupyter نمایش می‌دهیم.</p>
<p>این قابلیت برای پاسخ‌هایی که تیتر، فهرست، جدول یا قطعه‌کد دارند، بسیار کاربردی است.</p>
</div>

````python
from IPython.display import Markdown, display

display(Markdown(full_answer))
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>نمایش زندهٔ Markdown هنگام Streaming (اختیاری)</h3>
<p>اگر بخواهیم در خود Notebook محتوای دریافت‌شده را مرتباً به‌روزرسانی کنیم، می‌توانیم یک خروجی قابل به‌روزرسانی بسازیم. برای متن‌های خیلی طولانی، به‌روزرسانی در هر تکه ممکن است سنگین شود؛ در آن حالت بهتر است هر چند تکه یک‌بار خروجی را به‌روزرسانی کنیم.</p>
</div>

````python
stream = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a friendly Python teacher. Use simple English and Markdown."},
        {"role": "user", "content": "What is the OpenAI library?"},
    ],
    max_completion_tokens=300,
    stream=True,
)

handle = display(Markdown("Receiving response..."), display_id=True)
buffer = ""

for chunk in stream:
    if not chunk.choices:
        continue
    piece = chunk.choices[0].delta.content
    if piece:
        buffer += piece
        handle.update(Markdown(buffer))
    time.sleep(0.05)
````

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h2>۸. پارامترها و مفاهیم مهم دیگر</h2>

<h3>مدل (Model)</h3>
<p>مدل، هستهٔ هوش مصنوعی است که درخواست ما را دریافت و پردازش می‌کند و پاسخ می‌دهد. مدل‌های مختلف از نظر توانایی، سرعت، هزینه، پنجرهٔ زمینه و قابلیت‌های پشتیبانی‌شده با یکدیگر تفاوت دارند. هنگام ارسال درخواست به API، باید نام مدلی را مشخص کنیم که سرویس‌دهنده از آن پشتیبانی می‌کند.</p>

<p>شرکت‌های مختلف مدل‌های هوش مصنوعی متنوعی ارائه می‌دهند. برای مثال:</p>
<ul>
    <li><strong>OpenAI:</strong> مدل‌هایی مانند <code>GPT-4o</code> و <code>GPT-4.1</code></li>
    <li><strong>Google:</strong> مدل‌هایی مانند <code>Gemini 2.5 Flash</code> و <code>Gemini 2.5 Pro</code></li>
    <li><strong>Anthropic:</strong> مدل‌هایی مانند <code>Claude Sonnet</code> و <code>Claude Opus</code></li>
</ul>

<p>هر مدل از طریق سرویس‌دهندهٔ مربوط به خود قابل استفاده است. البته بعضی سرویس‌دهندگان واسطه، امکان دسترسی به مدل‌های شرکت‌های مختلف را از طریق یک API فراهم می‌کنند.</p>

<p>نام مدل را با استفاده از پارامتر <code>model</code> در درخواست API مشخص می‌کنیم. برای مثال: <code>model="gpt-4o"</code></p>

<h3>حداکثر خروجی (max_completion_tokens)</h3>
<p>سقف تعداد توکن‌هایی است که مدل می‌تواند برای تولید پاسخ استفاده کند. اگر مقدار را خیلی کم بگذاریم، ممکن است پاسخ ناقص شود؛ اگر مقدار بیشتری تعیین کنیم، مدل امکان تولید پاسخ طولانی‌تری خواهد داشت.</p>

<p>برای مثال، <code>max_completion_tokens=300</code> یعنی حداکثر ۳۰۰ توکن برای تولید پاسخ در نظر گرفته می‌شود.</p>
</div>

<div dir="rtl" style="text-align:right; line-height:1.9;">
<h3>Token چیست؟</h3>

<p>مدل‌های هوش مصنوعی متن را مستقیماً به شکل کلمه و جمله پردازش نمی‌کنند؛ بلکه ابتدا آن را به واحدهای کوچک‌تری به نام <strong>Token (توکن)</strong> تقسیم می‌کنند. هر توکن می‌تواند یک کلمه، بخشی از یک کلمه، یک عدد، علامت نگارشی یا حتی چند نویسه باشد.</p>

<p>برای مثال، جملهٔ زیر را در نظر بگیرید:</p>

<p style="text-align:center; direction:ltr; font-size:17px;"><strong>ChatGPT is useful.</strong></p>

<p>یک نمایش ساده و فرضی از تقسیم این جمله به توکن‌ها می‌تواند به شکل زیر باشد:</p>

<div dir="ltr" style="display:flex; justify-content:center; flex-wrap:wrap; gap:6px; margin:15px 0;">
    <span style="border:1px solid #4A90E2; background:#EAF3FF; color:#174A7E; padding:8px 12px; border-radius:6px;">Chat</span>
    <span style="border:1px solid #4A90E2; background:#EAF3FF; color:#174A7E; padding:8px 12px; border-radius:6px;">GPT</span>
    <span style="border:1px solid #4A90E2; background:#EAF3FF; color:#174A7E; padding:8px 12px; border-radius:6px;"> is</span>
    <span style="border:1px solid #4A90E2; background:#EAF3FF; color:#174A7E; padding:8px 12px; border-radius:6px;"> useful</span>
    <span style="border:1px solid #4A90E2; background:#EAF3FF; color:#174A7E; padding:8px 12px; border-radius:6px;">.</span>
</div>

<p>همان‌طور که می‌بینی، یک کلمه ممکن است به چند توکن تقسیم شود. بنابراین تعداد کلمات یک متن لزوماً با تعداد توکن‌های آن برابر نیست.</p>

<p>به‌طور کلی، در کار با API باید به دو نوع توکن توجه کنیم:</p>
<ul>
    <li><strong>Input Tokens:</strong> توکن‌های مربوط به اطلاعاتی که به مدل ارسال می‌کنیم؛ مانند پیام کاربر، دستورالعمل‌ها و تاریخچهٔ گفتگو.</li>
    <li><strong>Output Tokens:</strong> توکن‌هایی که مدل برای تولید پاسخ ایجاد می‌کند.</li>
</ul>

<p>برای مثال، اگر از مدل بپرسیم <code>What is Python?</code>، توکن‌های این سؤال جزو Input Tokens هستند و توکن‌هایی که مدل برای توضیح پایتون تولید می‌کند، جزو Output Tokens محسوب می‌شوند.</p>

<p>تعداد توکن‌ها به متن، زبان و Tokenizer مدل بستگی دارد. همچنین هزینهٔ استفاده از API معمولاً بر اساس تعداد توکن‌های ورودی و خروجی محاسبه می‌شود و نرخ آن بین مدل‌ها متفاوت است.</p>
</div>

````python
# Token usage information for a non-streaming response
usage_response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "Answer as a teacher."},
        {"role": "user", "content": "Explain the purpose of an API in one sentence."},
    ],
    max_completion_tokens=80,
)

print(usage_response.choices[0].message.content)

if usage_response.usage:
    print("Input tokens:", usage_response.usage.prompt_tokens)
    print("Output tokens:", usage_response.usage.completion_tokens)
    print("Total tokens:", usage_response.usage.total_tokens)
````

<h2>۹. مدیریت خطاها</h2>
<p>در کار با API ممکن است خطاهایی رخ دهند. بهتر است به‌جای اینکه برنامه با هر خطایی متوقف شود، علت‌های رایج را تشخیص بدهیم و پیام مناسبی نمایش دهیم.</p>
<table style="border-collapse:collapse; width:100%; text-align:right;">
<tr><th style="padding:6px; border-bottom:1px solid #bbb;">خطا</th><th style="padding:6px; border-bottom:1px solid #bbb;">علت‌های رایج</th><th style="padding:6px; border-bottom:1px solid #bbb;">اقدام اولیه</th></tr>
<tr><td style="padding:6px;">AuthenticationError</td><td style="padding:6px;">کلید نامعتبر، منقضی یا لغوشده</td><td style="padding:6px;">بررسی کلید و پروژه</td></tr>
<tr><td style="padding:6px;">RateLimitError</td><td style="padding:6px;">عبور از محدودیت نرخ یا سهمیه</td><td style="padding:6px;">کاهش دفعات درخواست و بررسی سهمیه</td></tr>
<tr><td style="padding:6px;">APITimeoutError</td><td style="padding:6px;">پاسخ در زمان تعیین‌شده نرسیده</td><td style="padding:6px;">بررسی شبکه و زمان انتظار</td></tr>
<tr><td style="padding:6px;">APIConnectionError</td><td style="padding:6px;">مشکل ارتباط، پراکسی یا شبکه</td><td style="padding:6px;">بررسی اتصال و آدرس API</td></tr>
<tr><td style="padding:6px;">BadRequestError</td><td style="padding:6px;">پارامتر یا ورودی نامعتبر</td><td style="padding:6px;">بررسی نام مدل و ساختار درخواست</td></tr>
</table>
</div>

````python
from openai import (
    AuthenticationError,
    RateLimitError,
    APITimeoutError,
    APIConnectionError,
    BadRequestError,
    APIStatusError,
)

try:
    test_response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "user", "content": "Hello! Just say the connection is working."},
        ],
        max_completion_tokens=50,
    )
    print(test_response.choices[0].message.content)

except AuthenticationError:
    print("Invalid API key or access denied.")

except RateLimitError:
    print("Check your API rate limit or quota.")

except APITimeoutError:
    print("The request timed out.")

except APIConnectionError:
    print("Connection failed. Check your network and BASE_URL.")

except BadRequestError as error:
    print("Invalid request. Check the model name and parameters.")
    print("Details:", error)

except APIStatusError as error:
    print("API error. Status code:", error.status_code)
    print("Details:", error)
````
